El círculo y la esfera son dos de las figuras geométricas más perfectas y estudiadas en la historia de las matemáticas. Desde la antigüedad, estas formas han fascinado a matemáticos, astrónomos y filósofos por su simetría y presencia en la naturaleza. En la era digital, comprender cómo calcular sus propiedades básicas —área, perímetro y volumen— se ha convertido en un ejercicio fundamental para los estudiantes de programación, especialmente aquellos que inician su camino en el desarrollo de software

 * Área del Círculo
Fórmula: A = π × r²
* Perímetro (Circunferencia) del Círculo
Fórmula: P = 2 × π × r o P = π × d
* Volumen de la Esfera
Fórmula: V = (4/3) × π × r³

```python
# ==========================================
# 📐 CÁLCULOS GEOMÉTRICOS SIMPLIFICADOS
# ==========================================
import math

def calcular(opcion):
    # Todas las funciones usan radio, así que lo pedimos una sola vez
    radio = float(input("📏 Ingresa el valor del radio: "))

    if opcion == "1":
        resultado = math.pi * radio ** 2
        print(f"✨ El Área es: {resultado:.2f}")

    elif opcion == "2":
        resultado = 2 * math.pi * radio
        print(f"✨ El Perímetro es: {resultado:.2f}")

    elif opcion == "3":
        resultado = (4/3) * math.pi * radio ** 3
        print(f"✨ El Volumen es: {resultado:.2f}")

# --- MENÚ DE USUARIO ---
while True:
    print("\n--- 🧮 MENÚ GEOMETRÍA ---")
    print("1. Área Círculo")
    print("2. Perímetro Círculo")
    print("3. Volumen Esfera")
    print("4. Salir")

    sel = input("👉 Elige (1-4): ")

    if sel == "4":
        print("¡Adiós! 👋")
        break
    elif sel in ["1", "2", "3"]:
        calcular(sel)
    else:
        print("⚠️ Opción inválida")

print("\nPrograma realizado por Oscar Flores NL = 19")1
```
